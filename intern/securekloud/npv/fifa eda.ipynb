```python
# loading required libraries
import pandas as pd
from matplotlib import pyplot as plt
import seaborn as sns
```

```python
fifa = pd.read_csv('players_15.csv')
fifa.head()
```

```python
# getting cols in table
for col in fifa.columns:
    print(col)
```

```python
# shape - gives no of rows and cols
fifa.shape
```

```python
# value_counts() - gives frequency of the categorical values
fifa['nationality'].value_counts()
```

```python
fifa['nationality'].value_counts()[0:10]
```

```python
fifa['nationality'].value_counts()[0:5]
```

```python
fifa['nationality'].value_counts()[0:5].keys() 
```

```python
plt.figure(figsize=(8,5))
plt.bar(list(fifa['nationality'].value_counts()[0:5].keys()), list(fifa['nationality'].value_counts()[0:5]), color='g')
plt.show()
```

```python
player_salary = fifa[['short_name', 'wage_eur']]
player_salary.head()
```

```python
player_salary=player_salary.sort_values(by=['wage_eur'], ascending=False)
player_salary.head()
```

```python
plt.figure(figsize=(8,5))
plt.bar(list(player_salary['short_name'])[0:5], list(player_salary['wage_eur'])[0:5], color=["r","b"])
plt.show()
```
