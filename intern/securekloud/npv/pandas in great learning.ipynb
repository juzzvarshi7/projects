```python
dataframe = pd.DataFrame({'Cricket':[1,2,np.nan,4,6,7,2,np.nan], 
                         'Baseball':[5, np.nan, np.nan, 5,7,2,4,5],
                         'Tennis': [1,2,3,4,5,6,7,8]})
dataframe
```

```python
dataframe.dropna()
# by default axis = 0
```

```python
# use axis = 1 for dropping columns with non values
dataframe.dropna(axis=1)
```

```python
dataframe.dropna(thresh=2)
# setting minimum number of null values to drop using threshold
```

```python
dataframe.fillna(value=0)
# replace empty values using value
```

```python
dataframe['Baseball'].fillna(value=dataframe['Baseball'].mean())
```

```python
# groupby method - groups rows together and perform aggregate functions
# create dataframe as given below
dat = {'CustID':['1001', '1001', '1002', '1002', '1003', '1003'],
      'CustName':['UIPat', 'DatROb', 'Goog', 'Chrysler', 'Ford', 'GM'],
      'ProfitinLakhs':[2005, 3245, 1234, 5544, 7543, 8623]}
dataframe = pd.DataFrame(dat)
dataframe
```

```python
CustID_grouped = dataframe.groupby('CustID')
CustID_grouped
```

```python
CustID_grouped.mean()
```

```python
CustID_grouped.median()
```

```python
CustID_grouped.std()
```

```python
CustID_grouped.min()
```

```python
CustID_grouped.max()
```

```python
CustID_grouped.count()
```

```python
CustID_grouped.describe()
# gives some imp stats 
```

```python
CustID_grouped.describe().transpose()
# also can be written as
#CustID_grouped.describe().T
```
