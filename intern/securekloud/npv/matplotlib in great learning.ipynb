```python
# line plot
# arange not arrange
import numpy as np
from matplotlib import pyplot as plt
# import matplotlib.pyplot as plt
x = np.arange(1,11)
x
```

```python
y = 2*x
y
```

```python
# build and show a plot
plt.plot(x,y)
plt.show()
```

```python
# add title and labels
plt.plot(x,y)
plt.title("line plot")
plt.xlabel("x-label")
plt.ylabel("y-label")
plt.show()
```

```python
# changing line attributes
# color not colour
plt.plot(x,y,color='g', linestyle=":", linewidth=2)
plt.show()
```

```python
# add two lines in the same plot
x = np.arange(1, 11)
y1 = 2*x
y2 = 3*x
plt.plot(x, y1, color='g', linestyle=':', linewidth=2)
plt.plot(x, y2, color='r', linestyle='-', linewidth=3)
plt.title("line plot")
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.grid(True)
plt.show()
```

```python
# adding sub plots
x=np.arange(1,11)
y1 = 2*x
y2=3*x
#plt.subplot(row, column, indexing)
plt.subplot(1,2,1)
plt.plot(x,y1, color='g', linestyle=':', linewidth=2)
plt.subplot(1,2,2)
plt.plot(x,y2, color='r', linestyle='-', linewidth=3)
plt.show()
```

```python
plt.subplot(2,1 ,1)
plt.plot(x,y1, color='orange', linestyle=':', linewidth=2)
plt.subplot(2,1,2)
plt.plot(x,y2, color='yellow', linestyle='-', linewidth=3)
plt.show()
```

```python
# bar plot
student = {"Bob":87, "Matt":56, "Sam":27}
names = list(student.keys())
values = list(student.values())
plt.bar(names, values)
plt.show()
```

```python
# horizontal bar graph
plt.barh(names, values, color='r')
plt.title('Bar Plot')
plt.xlabel("Marks")
plt.ylabel("Names")
plt.grid(True)
plt.show()
```

```python
# scatter plot and changing mark aesthetics
x = [10, 20, 30, 40, 50, 60, 70, 80, 90]
a = [8,1,7,2,0,3,7,3,2]
plt.scatter(x,a, marker="*", c='g', s=100)
plt.show()
```

```python
# adding two markers in the same plot
x = [10, 20, 30, 40, 50, 60, 70, 80, 90]
a = [8,1,7,2,0,3,7,3,2]
b = [7,2,5,6,9,1,4,5,3]
plt.scatter(x, a, marker="*", c='g', s = 100)
plt.scatter(x, b, marker = ".", c="r", s = 200)
plt.show()
```

```python
# adding subplots
x = [10, 20, 30, 40, 50, 60, 70, 80, 90]
a = [8,1,7,2,0,3,7,3,2]
b = [7,2,5,6,9,1,4,5,3]
plt.subplot(1,2,1)
plt.scatter(x,a,marker="*", c="g", s = 100)
plt.subplot(1,2,2)
plt.scatter(x,b, marker=".", c='r', s=200)
plt.show()
```

```python
# create box plot
one = [1,2,3,4,5,6,7,8,9]
two = [1,2,3,4,5,4,3,2,1]
three = [6,7,8,9,8,7,6,5,4]
data = list([one,two, three])
plt.boxplot(data)
plt.show()
# shows min max and median value
```

```python
# create violin plot
one = [1,2,3,4,5,6,7,8,9]
two = [1,2,3,4,5,4,3,2,1]
three = [6,7,8,9,8,7,6,5,4]
data = list([one,two, three])
plt.violinplot(data, showmedians=True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
fruit = ['Apple', 'Orange', 'Mango', 'Guava']
quantity = [67, 34, 100, 29]
plt.pie(quantity, labels=fruit, autopct='%0.1f%%', colors=['yellow', 'grey', 'blue', 'black'])
plt.show()
# autopct shows percentage in pie chart
```

```python
# donut chart
fruit = ['Apple', 'Orange', 'Mango', 'Guava']
quantity = [67, 34, 100, 29]

plt.pie(quantity, labels=fruit, radius = 2)
plt.pie([1], colors=['w'], radius = 1)
plt.show()
# w = color of the hole (white)
```
