```python
# TOTAL NUMBER OF MATCHES PLAYED ACCORDNING TO THE DATASET
data['id'].count()
```

```python
# NUMBER OF IPL SEASONS 
data['season'].unique()
```

```python
# WHICH IPL TEAM WON BY SCORING MAX RUNS
# idxmax gives the entire column related max of the row label
data.iloc[data['win_by_runs'].idxmax()]
```

```python
# WHICH IPL TEAM WON BY TAKING MAX WICKETS
data.iloc[data['win_by_wickets'].idxmax()]
```

```python
# WHICH IPL TEAM WON BY MIN WICKETS
data.iloc[data['win_by_wickets'].idxmin()]
```

```python
# WHICH SEASON HAS HIGHEST NO OF MATCHES
fig_dims = (20,4)
fig, ax = plt.subplots(figsize=fig_dims)
sns.countplot(x='season', ax=ax, data=data)
plt.show()
```

```python
# MOST SUCESSFUL IPL TEAM
# value_counts() - returns value in descending
# winner - row name
data1 = data.winner.value_counts()
sns.barplot(y=data1.index, x = data1)
plt.show()
```

```python
# PROBABILITY OF WINNING A MATCH IF TOSS IS LOST
probability_of_win = data['toss_winner'] == data['winner']
probability_of_win.groupby(probability_of_win).size()
```

```python
sns.countplot(x=probability_of_win)
```
