```python
#stacking methods  - vstack, hstack and column_stack()
# vstack - vertically stacking
m1 = np.array([10, 20, 30])
m2 = np.array([40, 50, 60])
np.vstack((m1, m2))
```

```python
#hstack - horizontally stacking
np.hstack((m1, m2))
```

```python
# column_stack - column stacking
np.column_stack((m1, m2))
```

```python
#intersection and differnce
# intersect1d - intersection of two matrix or common elements
o1 = np.array([10, 20, 30])
o2 = np.array([10, 50, 60])
np.intersect1d(o1, o2)
```

```python
#np.setdiff1d()
# unique to o1
np.setdiff1d(o1, o2)
```

```python
#unique to o2
np.setdiff1d(o2, o1)
```

```python
# total of arrays
#np.sum([n1, n2])
n1 = np.array([10, 20])
n2 = np.array([30, 40])
np.sum([n1,n2])
```

```python
# sum of elements along y axis
np.sum([n1, n2], axis=0)
```

```python
# sum of elements along x axis
np.sum([n1, n2], axis=1)
```

```python
# addition
n1 = np.array([10, 20, 30])
n1=n1+1
n1
```

```python
# mean
n1 = np.array([10, 20, 30, 40, 50, 60])
np.mean(n1)
```

```python
# standard deviation
n1 = np.array([1, 5, 3, 100, 4, 48])
np.std(n1)
```
